```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df=df.sort_values("date")
```

```python
df=df.set_index("date")
```

```python
plt.figure(figsize=(12,5))
sns.lineplot(
    data=df,
    x=df.index,
    y="open"
)
plt.title("Shopify Opening Price Over Time")
plt.xlabel("Date")
plt.ylabel("Opening Price")
plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.lineplot(
    data=df,
    x=df.index,
    y="close"
)
plt.title("Shopify Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.show()
```

```python
#Visualize OHLC price

plt.figure(figsize=(14,6))

sns.lineplot(data=df,x=df.index,y="open",label="open")
sns.lineplot(data=df,x=df.index,y="high",label="high")
sns.lineplot(data=df,x=df.index,y="low",label="low")
sns.lineplot(data=df,x=df.index,y="close",label="close")

plt.title("Shopify OHLC Price Over Time")
plt.xlabel("Date")
plt.ylabel("Price")


plt.show()
```

```python
 plt.figure(figsize=(14,5))

 sns.lineplot(data=df,x=df.index,y="volume")

 plt.title("Shopify Trading Volume Over Time")
 plt.xlabel("Date")
 plt.ylabel("Volume")
 plt.show()
```

```python
df.nlargest(5,"volume")
```

```python
df["daily price change"]=df["close"]-df["open"]
df["daily price change"]
```

```python
#daily volume vs daily price change

plt.figure(figsize=(10,6))

sns.scatterplot(data=df, x="volume", y="daily price change")

plt.title("trading Volume vs daily price change")
plt.xlabel("Volume")
plt.ylabel("Daily Price Change")

plt.show()
```

```python
df["daily price range"]=df["high"]-df["low"]
df["daily price range"]
```

```python
 #volume vs daily price range

 plt.figure(figsize=(10,6))

 sns.scatterplot(data=df, x="volume", y="daily price range")

 plt.title("trading Volume vs daily price range")
 plt.xlabel("Volume")
 plt.ylabel("Daily Price Range")

 plt.show()
```

```python
#SMA = sum of values in window / no.of values
```

```python
#calculate the 20-day SMA

df["SMA_20"]=df["close"].rolling(window=20).mean()
df["SMA_20"]
```

```python
#plot closing price + 20-day SMA

plt.figure(figsize=(14,6))

sns.lineplot(data=df,x="date",y="close",label="Closing Price")
sns.lineplot(data=df,x="date",y="SMA_20",label="20-day SMA")

plt.title("Shopify Closing Price vs 20-day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
#calculate the 50-day SMA

df["SMA_50"]=df["close"].rolling(window=50).mean()
df["SMA_50"]
```

```python
#plot closing price + 50-day SMA

plt.figure(figsize=(14,6))

sns.lineplot(data=df,x="date",y="close",label="Closing Price")
sns.lineplot(data=df,x="date",y="SMA_50",label="50-day SMA")

plt.title("Shopify Closing Price vs 50-day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
#plot close + both moving averages

plt.figure(figsize=(14,6))

sns.lineplot(data=df,x="date",y="close",label="Closing Price")
sns.lineplot(data=df,x="date",y="SMA_20",label="20-day SMA")
sns.lineplot(data=df,x="date",y="SMA_50",label="50-day SMA")

plt.title("Shopify Closing Price vs 20-day SMA vs 50-day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
#daily return %

df["daily return %"]= ((df["close"]-df["open"]) /df["open"]) *100
```

```python
df["daily return %"].head()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(data=df, x="daily return %", bins=20)

plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return %")
plt.ylabel("Frequency")

plt.show()
```

```python
#volatility-stock price or return fluctuate

df["absolute return"]=abs(df["daily return %"])
df["absolute return"]
```

```python
df.nlargest(10,"absolute return")
```

```python
plt.figure(figsize=(10,6))

sns.lineplot(data=df,x="date",y="daily return %")

plt.axhline(0,linestyle="--")
plt.title("Shopify Daily Return Over Time")
plt.xlabel("Date")
plt.ylabel("Daily Return %")

plt.show()
```
